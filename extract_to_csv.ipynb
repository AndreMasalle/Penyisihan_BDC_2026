# Extract data to csv

```python
# Import
from pathlib import Path
import csv
import pandas as pd
```

```python
# Config
# Relative path
TRAIN_DIR = "train"

# Absolute path
# TRAIN_DIR = r"C:\Users\username\......\train"

OUTPUT_CSV = "train.csv"

TRAIN_DIR = Path(TRAIN_DIR).expanduser().resolve()
OUTPUT_CSV = Path(OUTPUT_CSV).expanduser().resolve()

IMAGE_EXTENSIONS = {
    ".jpg",
    ".jpeg",
    ".png",
    ".bmp",
    ".webp"
}

if not TRAIN_DIR.exists() or not TRAIN_DIR.is_dir():
    raise FileNotFoundError(
        f"Folder dataset tidak ditemukan:\n{TRAIN_DIR}"
    )
```

```python
# Take data such label and file name from each image in the train folder
rows = []

for folder in sorted(TRAIN_DIR.iterdir()):

    if not folder.is_dir():
        continue

    label_number, label = folder.name.split("_", 1)

    prefix = label[0].upper()

    images = [
        img for img in folder.iterdir()
        if img.is_file() and img.suffix.lower() in IMAGE_EXTENSIONS
    ]

    def extract_number(path):
        match = re.search(r"\d+", path.stem)
        return int(match.group()) if match else float("inf")

    images.sort(key=extract_number)

    for index, image in enumerate(images, start=1):
        rows.append({
            "id": f"{prefix}_{index}",
            "label": label,
            "label_number": int(label_number),
            "label_folder_name": folder.name,
            "file_name": image.name
        })

print(f"Total data : {len(rows)}")
```

```python
# Export to CSV
with OUTPUT_CSV.open(
    mode="w",
    newline="",
    encoding="utf-8"
) as file:

    writer = csv.DictWriter(
        file,
        fieldnames=[
            "id",
            "label",
            "label_number",
            "label_folder_name",
            "file_name"
        ]
    )

    writer.writeheader()
    writer.writerows(rows)

print(f"CSV berhasil dibuat")
print(f"Path : {OUTPUT_CSV}")
```
